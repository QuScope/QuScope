```python
import numpy as np
import matplotlib.pyplot as plt

from quscope.quantum_ctem import (
    PerformanceBenchmark,
    BenchmarkVisualizer,
    quick_benchmark,
)

print("✅ Imports successful!")
```

## 1. Quick Benchmark Test

Test the quick benchmarking convenience function.

```python
# Quick benchmark for small grid
result = quick_benchmark(n_qubits=2)

print("\n📊 Quick Benchmark Results:")
print(f"   Grid size: {int(np.sqrt(result.pixels))}×{int(np.sqrt(result.pixels))}")
print(f"   Qubits: {result.n_qubits_x} + {result.n_qubits_y} = {result.n_qubits_x + result.n_qubits_y}")
print(f"   Encoding time: {result.encoding_time*1000:.3f} ms")
print(f"   Circuit depth: {result.circuit_depth}")
print(f"   Total gates: {result.total_gates}")
print(f"   Fidelity: {result.fidelity:.10f}")
print(f"   Round-trip error: {result.round_trip_error:.2e}")
print(f"   Est. IBM runtime: {result.estimated_runtime_ibm*1e6:.2f} µs")
print(f"   Est. IBM fidelity: {result.estimated_fidelity_ibm:.4f}")

assert result.fidelity > 0.999, "Fidelity too low!"
assert result.round_trip_error < 1e-8, "Error too high!"
print("\n✅ Quick benchmark passed!")
```

## 2. Scaling Analysis

Analyze performance across multiple grid sizes.

```python
benchmark = PerformanceBenchmark()

# Test small to medium grid sizes
results = benchmark.run_scaling_analysis(
    n_qubits_range=[2, 3, 4],
    num_trials=5
)

print("\n📈 Scaling Analysis Results:")
print("\n" + "="*80)
print(f"{'Grid Size':<12} {'Qubits':<8} {'Enc(ms)':<10} {'Depth':<8} {'Gates':<8} {'Fidelity':<12}")
print("="*80)

for r in results:
    grid_size = f"{int(np.sqrt(r.pixels))}×{int(np.sqrt(r.pixels))}"
    total_qubits = r.n_qubits_x + r.n_qubits_y
    enc_time = r.encoding_time * 1000
    print(f"{grid_size:<12} {total_qubits:<8} {enc_time:<10.3f} {r.circuit_depth:<8} {r.total_gates:<8} {r.fidelity:.10f}")

print("="*80)

# Verify all results meet quality standards
for r in results:
    assert r.fidelity > 0.999, f"Fidelity too low for {r.pixels} pixels"
    assert r.round_trip_error < 1e-8, f"Error too high for {r.pixels} pixels"

print("\n✅ All configurations meet quality standards!")
```

## 3. Optimization Method Comparison

Compare different state preparation methods.

```python
comparison = benchmark.compare_optimization_methods(n_qubits=3)

print("\n⚖️  Optimization Method Comparison (3 qubits):")
print("\n" + "="*70)
print(f"{'Method':<15} {'Prep Time(ms)':<15} {'Depth':<10} {'Gates':<10} {'Fidelity':<12}")
print("="*70)

for method, metrics in comparison.items():
    prep_time = metrics['preparation_time'] * 1000
    print(f"{method:<15} {prep_time:<15.3f} {metrics['depth']:<10} {metrics['gates']:<10} {metrics['fidelity']:.10f}")

print("="*70)
print("\n✅ Optimization comparison complete!")
```

## 4. Memory Profiling

Analyze memory usage across grid sizes.

```python
memory_data = benchmark.profile_memory_usage([2, 3, 4])

print("\n💾 Memory Profiling:")
print("\n" + "="*80)
print(f"{'Grid':<10} {'Classical(MB)':<15} {'Circuit(MB)':<15} {'Statevec(MB)':<15} {'Total(MB)':<12}")
print("="*80)

for i, pixels in enumerate(memory_data['pixels']):
    grid = f"{int(pixels)}×{int(pixels)}"
    classical = memory_data['classical_mb'][i]
    circuit = memory_data['circuit_mb'][i]
    statevec = memory_data['statevector_mb'][i]
    total = memory_data['total_mb'][i]
    print(f"{grid:<10} {classical:<15.3f} {circuit:<15.3f} {statevec:<15.3f} {total:<12.3f}")

print("="*80)
print("\n✅ Memory profiling complete!")
```

## 5. Hardware Cost Estimation

Estimate deployment costs for IBM quantum hardware.

```python
cost_data = benchmark.estimate_hardware_costs([2, 3, 4])

print("\n💰 Hardware Deployment Estimates (IBM Quantum):")
print("\n" + "="*70)
print(f"{'Grid':<10} {'Runtime(µs)':<15} {'Fidelity':<15} {'Cost Factor':<15}")
print("="*70)

for i, pixels in enumerate(cost_data['pixels']):
    grid = f"{int(pixels)}×{int(pixels)}"
    runtime = cost_data['runtime_seconds'][i] * 1e6
    fidelity = cost_data['estimated_fidelity'][i]
    # Cost factor = runtime * (1 - fidelity)
    cost_factor = runtime * (1 - fidelity)
    print(f"{grid:<10} {runtime:<15.2f} {fidelity:<15.4f} {cost_factor:<15.4f}")

print("="*70)
print("\n✅ Hardware cost estimation complete!")
```

Save benchmark data and generate report.

```python
# Save JSON results
benchmark.save_results(results, 'benchmark_results.json')
print("✅ Results saved to: benchmark_results.json")

# Generate markdown report
benchmark.generate_report(results, 'benchmark_report.md')
print("✅ Report saved to: benchmark_report.md")

# Show sample from report
with open('benchmark_report.md', 'r') as f:
    report = f.read()
    print("\n📄 Report Preview:")
    print("=" * 80)
    print(report[:500])
    print("...")
    print("=" * 80)
```

## 9. Summary Statistics

Calculate overall performance statistics.

```python
print("\n📊 Performance Summary:")
print("=" * 80)

# Extract metrics
enc_times = [r.encoding_time * 1000 for r in results]
dec_times = [r.decoding_time * 1000 for r in results]
fidelities = [r.fidelity for r in results]
errors = [r.round_trip_error for r in results]
depths = [r.circuit_depth for r in results]
gates = [r.total_gates for r in results]

print(f"\nTiming (ms):")
print(f"  Encoding: min={min(enc_times):.3f}, max={max(enc_times):.3f}, avg={np.mean(enc_times):.3f}")
print(f"  Decoding: min={min(dec_times):.3f}, max={max(dec_times):.3f}, avg={np.mean(dec_times):.3f}")

print(f"\nAccuracy:")
print(f"  Fidelity: min={min(fidelities):.10f}, avg={np.mean(fidelities):.10f}")
print(f"  Error: max={max(errors):.2e}, avg={np.mean(errors):.2e}")

print(f"\nCircuit Complexity:")
print(f"  Depth: min={min(depths)}, max={max(depths)}, avg={np.mean(depths):.1f}")
print(f"  Gates: min={min(gates)}, max={max(gates)}, avg={np.mean(gates):.1f}")

print("=" * 80)

# Validation
assert min(fidelities) > 0.999, "Fidelity below threshold!"
assert max(errors) < 1e-8, "Error above threshold!"

print("\n✅ All performance metrics meet requirements!")
```

# MoS2 Quantum vs Classical Showcase
This notebook runs the streamlined MoS2 workflow: builds a MoS2 patch, computes projected potentials (quantum converter vs abTEM), runs a classical multislice, and visualizes results.

```python
import matplotlib.pyplot as plt
from quscope.mos2_workflow import run_comparison, plot_structure_caxis, compare_projected_potentials, MicroscopeParams, get_interaction_constant, render_quantum_circuit
%matplotlib inline
```

```python
results = run_comparison(nx=3, ny=2, grid_size=256, pixel_size=0.1, voltage=200e3)
atoms = results['atoms']
Vq = results['V_quantum']
Vc = results['V_abtem']
Ic = results['I_classical']
```

```python
plot_structure_caxis(atoms, show=True)
compare_projected_potentials(Vq, Vc, sampling=0.1)
```

```python
mp = MicroscopeParams(acceleration_voltage=200e3, defocus=200.0, cs=1.0, aperture=50.0)
print('Microscope params:', mp)
sigma = get_interaction_constant(mp.acceleration_voltage)
print('Interaction constant σ:', sigma)
print(render_quantum_circuit(6))
```
